```python
target_community = {6, 11, 16, 23, 24, 25}
```

```python
def dice_rewire(G, community, budget_remove=2, budget_add=2):


    G = G.copy()

    # ---- Step 1: Remove internal edges ----
    internal_edges = [
        (u, v) for u, v in G.edges()
        if u in community and v in community
    ]

    remove_edges = random.sample(
        internal_edges,
        min(budget_remove, len(internal_edges))
    )

    G.remove_edges_from(remove_edges)

    # ---- Step 2: Add external edges ----
    outside_nodes = list(set(G.nodes()) - community)

    add_edges = []
    for _ in range(budget_add):
        u = random.choice(list(community))
        v = random.choice(outside_nodes)

        if not G.has_edge(u, v):
            add_edges.append((u, v))

    G.add_edges_from(add_edges)

    return G, remove_edges, add_edges
```

```python
G_dice, removed, added = dice_rewire(
    G,
    target_community,
    budget_remove=2,
    budget_add=2
)

print("Removed edges:", removed)
print("Added edges:", added)
```

```python
pos = nx.spring_layout(G, seed=42)

plt.figure(figsize=(12,5))

# BEFORE
plt.subplot(1,2,1)
nx.draw(G, pos, with_labels=True, node_color='black', font_color='white')
nx.draw_networkx_nodes(G, pos, nodelist=target_community, node_color='red')
plt.title("Before DICE")

# AFTER
plt.subplot(1,2,2)
nx.draw(G_dice, pos, with_labels=True, node_color='black', font_color='white')
nx.draw_networkx_nodes(G_dice, pos, nodelist=target_community, node_color='red')
plt.title("After DICE")

plt.savefig("dice_comparison.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import networkx as nx
import random
import numpy as np
from collections import defaultdict

# -------------------------------
# 1. Load Graph
# -------------------------------
def load_graph(file):
    G = nx.Graph()
    with open(file, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) == 2:
                u, v = map(int, parts)
                G.add_edge(u, v)
    return G


# -------------------------------
# 2. Louvain Communities (FIXED)
# -------------------------------
def get_communities(G):
    from networkx.algorithms.community import louvain_communities
    return list(louvain_communities(G, seed=42))


# -------------------------------
# 3. Centrality + Ranking
# -------------------------------
def compute_ranks(G):
    dc = nx.degree_centrality(G)
    cc = nx.closeness_centrality(G)
    bc = nx.betweenness_centrality(G)

    def rank_dict(d):
        return {node: rank+1 for rank, node in enumerate(sorted(d, key=d.get, reverse=True))}

    return dc, cc, bc, rank_dict(dc), rank_dict(cc), rank_dict(bc)


# -------------------------------
# 4. Core Community
# -------------------------------
def get_core_community(communities, rank_dc, rank_cc, rank_bc):
    best_comm = None
    best_score = float('inf')

    for comm in communities:
        avg_rank = np.mean([
            rank_dc[n] + rank_cc[n] + rank_bc[n]
            for n in comm
        ])
        if avg_rank < best_score:
            best_score = avg_rank
            best_comm = comm

    return best_comm


# -------------------------------
# 5. Mastermind
# -------------------------------
def get_mastermind(core_comm, rank_dc, rank_cc, rank_bc):
    scores = {}
    for n in core_comm:
        scores[n] = rank_dc[n] + rank_cc[n] + rank_bc[n]
    return min(scores, key=scores.get)


# -------------------------------
# 6. IC Model
# -------------------------------
def IC(G, seeds, p=0.1, simulations=100):
    total = 0
    for _ in range(simulations):
        active = set(seeds)
        new_active = set(seeds)

        while new_active:
            next_active = set()
            for node in new_active:
                for nbr in G.neighbors(node):
                    if nbr not in active and random.random() < p:
                        next_active.add(nbr)
            active |= next_active
            new_active = next_active

        total += len(active)

    return total / simulations


# -------------------------------
# 7. LT Model
# -------------------------------
def LT(G, seeds, simulations=100):
    total = 0

    for _ in range(simulations):
        thresholds = {n: random.random() for n in G.nodes()}
        active = set(seeds)

        changed = True
        while changed:
            changed = False
            for node in G.nodes():
                if node not in active:
                    neighbors = list(G.neighbors(node))
                    if not neighbors:
                        continue
                    frac = sum((nbr in active) for nbr in neighbors) / len(neighbors)
                    if frac >= thresholds[node]:
                        active.add(node)
                        changed = True

        total += len(active)

    return total / simulations


# -------------------------------
# 8. HERMES Iteration
# -------------------------------
def hermes_iteration(G, mastermind, core_comm, d=2):
    # External nodes
    external = [n for n in G.neighbors(mastermind) if n not in core_comm]

    if not external:
        return G

    # Rank external nodes by degree centrality
    dc = nx.degree_centrality(G)
    top_ext = sorted(external, key=lambda x: dc[x], reverse=True)[:d]

    # Remove edges from mastermind
    for n in top_ext:
        if G.has_edge(mastermind, n):
            G.remove_edge(mastermind, n)

    # Connect to least influential core nodes
    core_sorted = sorted(core_comm, key=lambda x: dc[x])

    for n in top_ext:
        for target in core_sorted[:2]:  # connect to 2 weak nodes
            if n != target:
                G.add_edge(n, target)

    return G


# ===============================
# MAIN EXECUTION
# ===============================

G = load_graph("facebook_fragment_333.txt")

# Communities
communities = get_communities(G)

# Centrality + ranks
dc, cc, bc, rank_dc, rank_cc, rank_bc = compute_ranks(G)

# Core community
core_comm = get_core_community(communities, rank_dc, rank_cc, rank_bc)
print("Core community size:", len(core_comm))

# Mastermind
mastermind = get_mastermind(core_comm, rank_dc, rank_cc, rank_bc)
print("Mastermind:", mastermind)

print("\nInitial Ranks:")
print("Degree Rank:", rank_dc[mastermind])
print("Closeness Rank:", rank_cc[mastermind])
print("Betweenness Rank:", rank_bc[mastermind])

# Initial IC, LT
print("\nInitial Influence:")
print("IC:", IC(G, [mastermind]))
print("LT:", LT(G, [mastermind]))

# -------------------------------
# HERMES ITERATIONS
# -------------------------------
for i in range(1, 4):
    G = hermes_iteration(G, mastermind, core_comm)

    print(f"\nAfter Iteration {i}:")
    print("IC:", IC(G, [mastermind]))
    print("LT:", LT(G, [mastermind]))

# Final ranks
dc, cc, bc, rank_dc, rank_cc, rank_bc = compute_ranks(G)

print("\nFinal Ranks:")
print("Degree Rank:", rank_dc[mastermind])
print("Closeness Rank:", rank_cc[mastermind])
print("Betweenness Rank:", rank_bc[mastermind])
```
