```python
import networkx as nx
import pandas as pd

def apply_roam_and_check_centrality(graph_filepath, budget_b):
    """
    Applies the ROAM algorithm to a given network and checks the reduction
    in the mastermind node's centrality.

    Args:
        graph_filepath (str): Path to the text file containing the graph's edge list.
        budget_b (int): The budget 'b' for the ROAM algorithm.
    """
    try:
        # 1. Load the graph
        # The file contains an edge list, one edge per line, with nodes separated by a tab.
        # Assuming node names are integers, read them as such.
        G = nx.read_edgelist(graph_filepath, create_using=nx.Graph(), nodetype=int)
        print(f"Graph loaded successfully from {graph_filepath} with {G.number_of_nodes()} nodes and {G.number_of_edges()} edges.")

        # Calculate initial centrality measures
        print("\nCalculating initial centrality measures...")
        initial_degree_centrality = nx.degree_centrality(G)
        initial_closeness_centrality = nx.closeness_centrality(G)
        initial_betweenness_centrality = nx.betweenness_centrality(G)

        # Convert to pandas DataFrame for easier ranking and analysis
        df_initial_centrality = pd.DataFrame({
            'degree': pd.Series(initial_degree_centrality),
            'closeness': pd.Series(initial_closeness_centrality),
            'betweenness': pd.Series(initial_betweenness_centrality)
        })

        # Rank nodes for each centrality measure (lower rank number means higher centrality)
        df_initial_centrality['degree_rank'] = df_initial_centrality['degree'].rank(ascending=False, method='min')
        df_initial_centrality['closeness_rank'] = df_initial_centrality['closeness'].rank(ascending=False, method='min')
        df_initial_centrality['betweenness_rank'] = df_initial_centrality['betweenness'].rank(ascending=False, method='min')

        # Find the mastermind (source node v†)
        # [cite_start]"The source node is assumed to be the one with the lowest sum of centrality rankings (ties are broken uniformly at random)." [cite: 1]
        df_initial_centrality['sum_ranks'] = df_initial_centrality['degree_rank'] + \
                                              df_initial_centrality['closeness_rank'] + \
                                              df_initial_centrality['betweenness_rank']

        mastermind_node = df_initial_centrality['sum_ranks'].idxmin()
        print(f"\nIdentified Mastermind node (v†): {mastermind_node}")

        initial_mastermind_centralities = df_initial_centrality.loc[mastermind_node]
        print("\n--- Initial Centrality Measures for Mastermind Node ---")
        print(f"Degree Centrality: {initial_mastermind_centralities['degree']:.4f}")
        print(f"Closeness Centrality: {initial_mastermind_centralities['closeness']:.4f}")
        print(f"Betweenness Centrality: {initial_mastermind_centralities['betweenness']:.4f}")
        print("\n--- Initial Ranks for Mastermind Node ---")
        print(f"Degree Rank: {initial_mastermind_centralities['degree_rank']:.0f}")
        print(f"Closeness Rank: {initial_mastermind_centralities['closeness_rank']:.0f}")
        print(f"Betweenness Rank: {initial_mastermind_centralities['betweenness_rank']:.0f}")

        # [cite_start]Apply the ROAM algorithm once [cite: 15, 16]
        print(f"\nApplying ROAM algorithm for mastermind node {mastermind_node} with budget b = {budget_b}...")
        G_modified = G.copy()

        mastermind_neighbors = list(G_modified.neighbors(mastermind_node))

        if not mastermind_neighbors:
            print(f"Mastermind node {mastermind_node} has no neighbors. ROAM cannot be applied.")
            return

        # [cite_start]Step 1: Choose v0 (neighbour of v† with most connections) [cite: 31]
        # We need degrees of current neighbors to make this choice
        neighbor_degrees = {n: G_modified.degree(n) for n in mastermind_neighbors}
        v0 = max(neighbor_degrees, key=neighbor_degrees.get)
        print(f"Chosen neighbor v0 (neighbor of v† with most connections): {v0}")

        # [cite_start]Step 2: Choose b-1 nodes of choice (neighbours of v† but not of v0 with least connections) [cite: 16, 31]
        potential_v0_new_connections = [
            n for n in mastermind_neighbors
            if n != v0 and not G_modified.has_edge(v0, n)
        ]

        # Sort these potential connections by their degree (least connections first)
        potential_v0_new_connections_sorted = sorted(
            potential_v0_new_connections,
            key=lambda x: G_modified.degree(x)
        )

        # Select b-1 nodes, ensuring not to exceed available unique neighbors
        num_connections_to_add = min(budget_b - 1, len(potential_v0_new_connections_sorted))
        nodes_to_connect_v0 = potential_v0_new_connections_sorted[:num_connections_to_add]
        print(f"Nodes to connect v0 to (b-1 neighbors of v† with least connections, not already connected to v0): {nodes_to_connect_v0}")

        # [cite_start]As per the paper's analogy with Facebook, Step 2 (adding edges) is done before Step 1 (removing edge) [cite: 36]
        for node in nodes_to_connect_v0:
            if not G_modified.has_edge(v0, node):
                G_modified.add_edge(v0, node)
                print(f"  Added edge: ({v0}, {node})")
            else:
                print(f"  Edge ({v0}, {node}) already exists, skipping addition.")

        # [cite_start]Apply Step 1 (remove the link between v† and v0) [cite: 15]
        if G_modified.has_edge(mastermind_node, v0):
            G_modified.remove_edge(mastermind_node, v0)
            print(f"  Removed edge: ({mastermind_node}, {v0})")
        else:
            print(f"  No edge between {mastermind_node} and {v0} to remove.")

        # Calculate new centrality measures after ROAM
        print("\nCalculating new centrality measures after ROAM...")
        new_degree_centrality = nx.degree_centrality(G_modified)
        new_closeness_centrality = nx.closeness_centrality(G_modified)
        new_betweenness_centrality = nx.betweenness_centrality(G_modified)

        # Check if the centrality of the mastermind has reduced
        print("\n--- Centrality Comparison (Mastermind Node) ---")
        print(f"Mastermind Node: {mastermind_node}")

        # Degree Centrality
        print(f"\nInitial Degree Centrality: {initial_mastermind_centralities['degree']:.4f}")
        print(f"New Degree Centrality: {new_degree_centrality.get(mastermind_node, 0):.4f}")
        if new_degree_centrality.get(mastermind_node, 0) < initial_mastermind_centralities['degree']:
            print("  -> Degree Centrality REDUCED.")
        else:
            print("  -> Degree Centrality NOT REDUCED (or increased/remained same).")

        # Closeness Centrality
        print(f"\nInitial Closeness Centrality: {initial_mastermind_centralities['closeness']:.4f}")
        print(f"New Closeness Centrality: {new_closeness_centrality.get(mastermind_node, 0):.4f}")
        if new_closeness_centrality.get(mastermind_node, 0) < initial_mastermind_centralities['closeness']:
            print("  -> Closeness Centrality REDUCED.")
        else:
            print("  -> Closeness Centrality NOT REDUCED (or increased/remained same).")

        # Betweenness Centrality
        print(f"\nInitial Betweenness Centrality: {initial_mastermind_centralities['betweenness']:.4f}")
        print(f"New Betweenness Centrality: {new_betweenness_centrality.get(mastermind_node, 0):.4f}")
        if new_betweenness_centrality.get(mastermind_node, 0) < initial_mastermind_centralities['betweenness']:
            print("  -> Betweenness Centrality REDUCED.")
        else:
            print("  -> Betweenness Centrality NOT REDUCED (or increased/remained same).")

    except FileNotFoundError:
        print(f"Error: The file '{graph_filepath}' was not found. Please ensure it's in the correct directory.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# To run the analysis:
# Make sure 'nodes.txt' is in the same directory as your Python script
# And you have networkx and pandas installed (`pip install networkx pandas`)
apply_roam_and_check_centrality('nodes.txt',3)
```

```python
import networkx as nx
import random
import community as community_louvain

random.seed(42)

# ---------------------------------------------------
# Centrality + Ranking
# ---------------------------------------------------

def compute_centrality_and_ranks(G):

    dc = nx.degree_centrality(G)
    cc = nx.closeness_centrality(G)
    bc = nx.betweenness_centrality(G)

    def rank_nodes(metric):

        nodes = list(metric.keys())
        random.shuffle(nodes)  # break ties randomly

        sorted_nodes = sorted(nodes, key=lambda x: metric[x], reverse=True)

        return {node: rank+1 for rank, node in enumerate(sorted_nodes)}

    dc_rank = rank_nodes(dc)
    cc_rank = rank_nodes(cc)
    bc_rank = rank_nodes(bc)

    return dc_rank, cc_rank, bc_rank


# ---------------------------------------------------
# Identify Mastermind (same logic as C3R)
# ---------------------------------------------------

def find_mastermind(G, partition):

    dc_rank, cc_rank, bc_rank = compute_centrality_and_ranks(G)

    communities = {}

    for node, com in partition.items():
        communities.setdefault(com, []).append(node)

    # Average rank for each community
    community_score = {}

    for com, nodes in communities.items():

        avg_rank = sum(
            dc_rank[n] + cc_rank[n] + bc_rank[n]
            for n in nodes
        ) / len(nodes)

        community_score[com] = avg_rank

    # core community
    core_community = min(community_score, key=community_score.get)

    core_nodes = communities[core_community]

    # mastermind = best ranked node in core community
    mastermind = min(
        core_nodes,
        key=lambda n: dc_rank[n] + cc_rank[n] + bc_rank[n]
    )

    return mastermind, core_nodes


# ---------------------------------------------------
# HERMES Rewiring
# ---------------------------------------------------

def hermes(G, mastermind, core_nodes, partition, budget=3):

    G = G.copy()

    neighbors = list(G.neighbors(mastermind))

    # External neighbors (outside mastermind community)
    external_neighbors = [
        n for n in neighbors
        if partition[n] != partition[mastermind]
    ]

    if len(external_neighbors) == 0:
        return G

    # Centralities to determine influence
    dc = nx.degree_centrality(G)
    cc = nx.closeness_centrality(G)
    bc = nx.betweenness_centrality(G)

    influence = {
        n: dc[n] + cc[n] + bc[n]
        for n in external_neighbors
    }

    # Top influential external neighbors
    top_nodes = sorted(
        external_neighbors,
        key=lambda x: influence[x],
        reverse=True
    )[:budget]

    for node in top_nodes:

        # Step 1: Remove mastermind edge
        if G.has_edge(mastermind, node):
            G.remove_edge(mastermind, node)

        # Step 2: Find weakest node in mastermind community
        weak_core_node = min(
            core_nodes,
            key=lambda x: G.degree(x)
        )

        # Step 3: Connect external node to weak core node
        if node != weak_core_node and not G.has_edge(node, weak_core_node):
            G.add_edge(node, weak_core_node)

    return G


# ---------------------------------------------------
# Experiment Runner
# ---------------------------------------------------

def run_hermes(filepath, budget=3, iterations=5):

    G = nx.read_edgelist(filepath, nodetype=int)

    print("\nGraph Loaded")
    print("Nodes:", G.number_of_nodes())
    print("Edges:", G.number_of_edges())

    # Detect communities
    partition = community_louvain.best_partition(G)

    mastermind, core_nodes = find_mastermind(G, partition)

    dc_rank, cc_rank, bc_rank = compute_centrality_and_ranks(G)

    print("\nMastermind:", mastermind)

    print("\nInitial Mastermind Ranks")
    print("Degree Rank:", dc_rank[mastermind])
    print("Closeness Rank:", cc_rank[mastermind])
    print("Betweenness Rank:", bc_rank[mastermind])

    # Run heuristic
    for i in range(iterations):

        print("\nIteration", i+1)

        partition = community_louvain.best_partition(G)

        G = hermes(G, mastermind, core_nodes, partition, budget)

    # Final centrality ranks
    dc_rank, cc_rank, bc_rank = compute_centrality_and_ranks(G)

    print("\nFinal Mastermind Ranks")
    print("Degree Rank:", dc_rank[mastermind])
    print("Closeness Rank:", cc_rank[mastermind])
    print("Betweenness Rank:", bc_rank[mastermind])


# ---------------------------------------------------

run_hermes("facebook_fragment_333.txt", budget=5, iterations=20)
```

#### C3R

```python
import networkx as nx
import random
import numpy as np
from networkx.algorithms.community import louvain_communities

random.seed(42)
np.random.seed(42)


# ---------------------------------------------------
# Rank nodes inside a community (random tie breaking)
# ---------------------------------------------------
def rank_nodes(G, nodes):

    sub = G.subgraph(nodes)

    dc = nx.degree_centrality(sub)
    cc = nx.closeness_centrality(sub)
    bc = nx.betweenness_centrality(sub)

    def rank_metric(metric):

        items = list(metric.items())

        random.shuffle(items)  # break ties randomly

        sorted_nodes = sorted(items, key=lambda x: x[1], reverse=True)

        return {node: i+1 for i, (node, _) in enumerate(sorted_nodes)}

    dc_rank = rank_metric(dc)
    cc_rank = rank_metric(cc)
    bc_rank = rank_metric(bc)

    return dc_rank, cc_rank, bc_rank


# ---------------------------------------------------
# Find core community
# ---------------------------------------------------
def find_core_community(G, communities):

    best_avg = float("inf")
    core = None

    for community in communities:

        dc_rank, cc_rank, bc_rank = rank_nodes(G, community)

        total = 0

        for node in community:
            total += dc_rank[node] + cc_rank[node] + bc_rank[node]

        avg_rank = total / len(community)

        if avg_rank < best_avg:
            best_avg = avg_rank
            core = community

    return core


# ---------------------------------------------------
# Find mastermind
# ---------------------------------------------------
def find_mastermind(G, core):

    dc_rank, cc_rank, bc_rank = rank_nodes(G, core)

    scores = {}

    for node in core:

        scores[node] = dc_rank[node] + cc_rank[node] + bc_rank[node]

    mastermind = min(scores, key=scores.get)

    return mastermind


# ---------------------------------------------------
# Find weakest node in a community
# ---------------------------------------------------
def weakest_node(G, community):

    sub = G.subgraph(community)

    dc = nx.degree_centrality(sub)

    weakest = min(dc, key=dc.get)

    return weakest


# ---------------------------------------------------
# Compute global centrality ranks
# ---------------------------------------------------
def global_ranks(G):

    dc = nx.degree_centrality(G)
    cc = nx.closeness_centrality(G)
    bc = nx.betweenness_centrality(G)

    def rank(metric):

        items = list(metric.items())

        random.shuffle(items)

        sorted_nodes = sorted(items, key=lambda x: x[1], reverse=True)

        return {node: i+1 for i,(node,_) in enumerate(sorted_nodes)}

    return rank(dc), rank(cc), rank(bc)


# ---------------------------------------------------
# C3R rewiring
# ---------------------------------------------------
def c3r_rewire(G, mastermind, core, communities, d):

    G = G.copy()

    external_nodes = [n for n in G.neighbors(mastermind) if n not in core]

    if len(external_nodes) == 0:
        return G

    # global centrality ranking
    dc_rank, cc_rank, bc_rank = global_ranks(G)

    scores = {}

    for n in external_nodes:
        scores[n] = dc_rank[n] + cc_rank[n] + bc_rank[n]

    # top d influential external nodes
    top_external = sorted(scores, key=scores.get)[:d]

    for Oi in top_external:

        # find Oi community
        Oi_comm = None

        for comm in communities:
            if Oi in comm:
                Oi_comm = comm
                break

        if Oi_comm is None:
            continue

        Wi = weakest_node(G, Oi_comm)
        MWi = weakest_node(G, core)

        # remove mastermind → Oi
        if G.has_edge(mastermind, Oi):
            G.remove_edge(mastermind, Oi)

        # add chain edges
        if not G.has_edge(Oi, Wi):
            G.add_edge(Oi, Wi)

        if not G.has_edge(Wi, MWi):
            G.add_edge(Wi, MWi)

        if not G.has_edge(MWi, mastermind):
            G.add_edge(MWi, mastermind)

    return G


# ---------------------------------------------------
# Main C3R Experiment
# ---------------------------------------------------
def run_c3r(filepath, iterations, d):

    G = nx.read_edgelist(filepath, nodetype=int)

    print("Graph loaded.")
    print("Nodes:", G.number_of_nodes())
    print("Edges:", G.number_of_edges())

    # -------- INITIAL COMMUNITY --------

    communities = louvain_communities(G, seed=42)

    core = find_core_community(G, communities)

    mastermind = find_mastermind(G, core)

    print("\nCore community size:", len(core))
    print("Mastermind:", mastermind)

    # initial centrality ranks
    dc_rank, cc_rank, bc_rank = global_ranks(G)

    print("\nInitial Mastermind Ranks")
    print("Degree Rank:", dc_rank[mastermind])
    print("Closeness Rank:", cc_rank[mastermind])
    print("Betweenness Rank:", bc_rank[mastermind])

    # -------- ITERATIONS --------

    for i in range(iterations):

        print("\nIteration", i+1)

        communities = louvain_communities(G, seed=42)

        core = find_core_community(G, communities)

        G = c3r_rewire(G, mastermind, core, communities, d)

    # -------- FINAL CENTRALITY --------

    dc_rank_f, cc_rank_f, bc_rank_f = global_ranks(G)

    print("\nFinal Mastermind Ranks")
    print("Degree Rank:", dc_rank_f[mastermind])
    print("Closeness Rank:", cc_rank_f[mastermind])
    print("Betweenness Rank:", bc_rank_f[mastermind])

    return G


# ---------------------------------------------------
# Run
# ---------------------------------------------------
run_c3r("facebook_fragment_333.txt", iterations=20, d=5)
```

```python
import networkx as nx
import random
import numpy as np

random.seed(42)
np.random.seed(42)

# ---------------------------------------------------
# Independent Cascade Model
# ---------------------------------------------------
def independent_cascade(G, source, p=0.15):
    active = set([source])
    newly_active = set([source])

    while newly_active:
        next_active = set()
        for node in newly_active:
            for neighbor in G.neighbors(node):
                if neighbor not in active:
                    if random.random() <= p:
                        next_active.add(neighbor)

        newly_active = next_active
        active |= newly_active

    return len(active)


# ---------------------------------------------------
# Linear Threshold Model
# ---------------------------------------------------

def linear_threshold(G, source, delta=0.05): # resistant
    active = set([source])
    thresholds = {node: random.random() for node in G.nodes()}
    weights = 1 / np.array([G.degree(n) for n in G.nodes()])

    while True:
        new_active = set()

        for node in G.nodes():
            if node not in active:
                influence = sum(
                    1/G.degree(neigh)
                    for neigh in G.neighbors(node)
                    if neigh in active
                )

                if influence >= thresholds[node]:
                    new_active.add(node)
                else:
                    thresholds[node] += delta   # 🔥 LT Increase

        if not new_active:
            break

        active |= new_active

    return len(active)

def estimate_influence(G, source, model="IC",
                       simulations=500,
                       delta=0.05):

    spreads = []

    for _ in range(simulations):

        if model == "IC":
            spread = independent_cascade(G, source)

        elif model == "LT":
            spread = linear_threshold(G, source)

        elif model == "LT_RESIST":
            spread = linear_threshold(
                        G,
                        source,
                        delta=delta   # threshold growth rate
                     )

        spreads.append(spread)

    return np.mean(spreads)
# ---------------------------------------------------
# Rank Helper
# ---------------------------------------------------
def compute_centrality_and_ranks(G):
    degree = nx.degree_centrality(G)
    closeness = nx.closeness_centrality(G)
    betweenness = nx.betweenness_centrality(G)

    # Ranking
    degree_rank = {
        node: rank+1 for rank, node in
        enumerate(sorted(degree, key=degree.get, reverse=True))
    }

    closeness_rank = {
        node: rank+1 for rank, node in
        enumerate(sorted(closeness, key=closeness.get, reverse=True))
    }

    betweenness_rank = {
        node: rank+1 for rank, node in
        enumerate(sorted(betweenness, key=betweenness.get, reverse=True))
    }

    return degree, closeness, betweenness, degree_rank, closeness_rank, betweenness_rank

# ---------------------------------------------------
# CCCR Implementation (Community Cascade Rerouting)
# ---------------------------------------------------
from networkx.algorithms.community import greedy_modularity_communities
from networkx.algorithms.community.quality import modularity

def cccr(G, mastermind, budget=3):

    G = G.copy()

    # --- Detect communities ---
    communities = list(greedy_modularity_communities(G))
    node_to_comm = {}
    for i, comm in enumerate(communities):
        for node in comm:
            node_to_comm[node] = i

    core_comm_id = node_to_comm[mastermind]
    core_comm = communities[core_comm_id]

    # --- Compute centralities ---
    bet = nx.betweenness_centrality(G)
    deg = dict(G.degree())

    # --- Define weakness score (low degree + low betweenness) ---
    def weakness(n):
        return bet[n] + deg[n]

    # --- Identify riskiest external neighbor ---
    external_neighbors = [
        n for n in G.neighbors(mastermind)
        if node_to_comm[n] != core_comm_id
    ]

    if not external_neighbors:
        return G

    R = max(external_neighbors, key=lambda n: bet[n])

    # --- Remove direct mastermind link ---
    if G.has_edge(mastermind, R):
        G.remove_edge(mastermind, R)

    # --- Connect weakest core node to R ---
    core_weak = min(core_comm, key=weakness)

    if not G.has_edge(core_weak, R):
        G.add_edge(core_weak, R)

    # --- Chain through other communities ---
    other_comms = [
        communities[i] for i in range(len(communities))
        if i != core_comm_id
    ]

    prev = R
    count = 0

    for comm in other_comms:

        weakest_node = min(comm, key=weakness)

        if not G.has_edge(prev, weakest_node):
            G.add_edge(prev, weakest_node)

        prev = weakest_node
        count += 1

        if count >= budget:
            break

    return G


# ---------------------------------------------------
# Print Mastermind Metrics
# ---------------------------------------------------
from networkx.algorithms.community.quality import modularity

def print_metrics(G, mastermind, stage):

    degree, closeness, betweenness, d_rank, c_rank, b_rank = compute_centrality_and_ranks(G)

    ic_val = estimate_influence(G, mastermind, model="IC")
    lt_val = estimate_influence(G, mastermind, model="LT")

    communities = list(greedy_modularity_communities(G))
    mod_val = modularity(G, communities)

    print("\n====================================")
    print(stage)
    print("====================================")
    print("Mastermind:", mastermind)

    print("\nCentrality Values:")
    print("Degree:", degree[mastermind])
    print("Closeness:", closeness[mastermind])
    print("Betweenness:", betweenness[mastermind])

    print("\nCentrality Ranks:")
    print("Degree Rank:", d_rank[mastermind])
    print("Closeness Rank:", c_rank[mastermind])
    print("Betweenness Rank:", b_rank[mastermind])

    print("\nInfluence Spread:")
    print("IC Influence:", ic_val)
    print("LT Influence:", lt_val)

    print("\nStructural Property:")
    print("Modularity:", mod_val)


def run_experiment(filepath, budget=3, executions=5, simulations=500):

    G = nx.read_edgelist(filepath, nodetype=int)

    # Identify mastermind
    degree, closeness, betweenness, d_rank, c_rank, b_rank = compute_centrality_and_ranks(G)

    sum_ranks = {
        node: d_rank[node] + c_rank[node] + b_rank[node]
        for node in G.nodes()
    }

    mastermind = min(sum_ranks, key=sum_ranks.get)

    print("Mastermind:", mastermind)

    # ---- INITIAL INFLUENCE ----
    ic_initial = estimate_influence(G, mastermind, model="IC", simulations=simulations)
    lt_initial = estimate_influence(G, mastermind, model="LT", simulations=simulations)

    print("\nInitial IC:", ic_initial)
    print("Initial LT:", lt_initial)

    ic_values = [ic_initial]
    lt_values = [lt_initial]

    # ---- Apply ROAM repeatedly ----
    for i in range(executions):
        G = cccr(G, mastermind, budget)

        ic_val = estimate_influence(G, mastermind, model="IC", simulations=simulations)
        lt_val = estimate_influence(G, mastermind, model="LT", simulations=simulations)

        ic_values.append(ic_val)
        lt_values.append(lt_val)

        print(f"\nAfter execution {i+1}:")
        print("IC:", ic_val)
        print("LT:", lt_val)

    # ---- Compute changes ----
    ic_total_change = ic_values[-1] - ic_initial
    lt_total_change = lt_values[-1] - lt_initial

    ic_avg_change = ic_total_change / executions
    lt_avg_change = lt_total_change / executions

    print("\n====================================")
    print("SUMMARY AFTER", executions, "EXECUTIONS")
    print("====================================")

    print("Total IC Change:", ic_total_change)
    print("Average IC Change per execution:", ic_avg_change)

    print("Total LT Change:", lt_total_change)
    print("Average LT Change per execution:", lt_avg_change)
    
run_experiment("nodes.txt", budget=3, executions=3)

# executions here means - “If I repeatedly apply ROAM 20 times to structurally dilute the mastermind's neighbourhood,
# how does their influence change over time?”
```

```python
import networkx as nx
import random
import numpy as np

random.seed(42)
np.random.seed(42)

# ---------------------------------------------------
# Independent Cascade Model
# ---------------------------------------------------
def independent_cascade(G, source, p=0.15):
    active = set([source])
    newly_active = set([source])

    while newly_active:
        next_active = set()
        for node in newly_active:
            for neighbor in G.neighbors(node):
                if neighbor not in active:
                    if random.random() <= p:
                        next_active.add(neighbor)

        newly_active = next_active
        active |= newly_active

    return len(active)


# ---------------------------------------------------
# Linear Threshold Model
# ---------------------------------------------------

def linear_threshold(G, source, delta=0.05): # resistant
    active = set([source])
    thresholds = {node: random.random() for node in G.nodes()}
    weights = 1 / np.array([G.degree(n) for n in G.nodes()])

    while True:
        new_active = set()

        for node in G.nodes():
            if node not in active:
                influence = sum(
                    1/G.degree(neigh)
                    for neigh in G.neighbors(node)
                    if neigh in active
                )

                if influence >= thresholds[node]:
                    new_active.add(node)
                else:
                    thresholds[node] += delta   # 🔥 LT Increase

        if not new_active:
            break

        active |= new_active

    return len(active)

def estimate_influence(G, source, model="IC",
                       simulations=500,
                       delta=0.05):

    spreads = []

    for _ in range(simulations):

        if model == "IC":
            spread = independent_cascade(G, source)

        elif model == "LT":
            spread = linear_threshold(G, source)

        elif model == "LT_RESIST":
            spread = linear_threshold(
                        G,
                        source,
                        delta=delta   # threshold growth rate
                     )

        spreads.append(spread)

    return np.mean(spreads) / G.number_of_nodes()
# ---------------------------------------------------
# Rank Helper
# ---------------------------------------------------
def compute_centrality_and_ranks(G):
    degree = nx.degree_centrality(G)
    closeness = nx.closeness_centrality(G)
    betweenness = nx.betweenness_centrality(G)

    # Ranking
    degree_rank = {
        node: rank+1 for rank, node in
        enumerate(sorted(degree, key=degree.get, reverse=True))
    }

    closeness_rank = {
        node: rank+1 for rank, node in
        enumerate(sorted(closeness, key=closeness.get, reverse=True))
    }

    betweenness_rank = {
        node: rank+1 for rank, node in
        enumerate(sorted(betweenness, key=betweenness.get, reverse=True))
    }

    return degree, closeness, betweenness, degree_rank, closeness_rank, betweenness_rank

# ---------------------------------------------------
# CCCR Implementation (Community Cascade Rerouting)
# ---------------------------------------------------
from networkx.algorithms.community import greedy_modularity_communities
from networkx.algorithms.community.quality import modularity

def cccr(G, mastermind, budget=3):

    G = G.copy()

    # --- Detect communities ---
    communities = list(greedy_modularity_communities(G))
    node_to_comm = {}
    for i, comm in enumerate(communities):
        for node in comm:
            node_to_comm[node] = i

    core_comm_id = node_to_comm[mastermind]
    core_comm = communities[core_comm_id]

    # --- Compute centralities ---
    bet = nx.betweenness_centrality(G)
    deg = dict(G.degree())

    # --- Define weakness score (low degree + low betweenness) ---
    def weakness(n):
        return bet[n] + deg[n]

    # --- Identify riskiest external neighbor ---
    external_neighbors = [
        n for n in G.neighbors(mastermind)
        if node_to_comm[n] != core_comm_id
    ]

    if not external_neighbors:
        return G

    R = max(external_neighbors, key=lambda n: bet[n])

    # --- Remove direct mastermind link ---
    if G.has_edge(mastermind, R):
        G.remove_edge(mastermind, R)

    # --- Connect weakest core node to R ---
    core_weak = min(core_comm, key=weakness)

    if not G.has_edge(core_weak, R):
        G.add_edge(core_weak, R)

    # --- Chain through other communities ---
    other_comms = [
        communities[i] for i in range(len(communities))
        if i != core_comm_id
    ]

    prev = R
    count = 0

    for comm in other_comms:

        weakest_node = min(comm, key=weakness)

        if not G.has_edge(prev, weakest_node):
            G.add_edge(prev, weakest_node)

        prev = weakest_node
        count += 1

        if count >= budget:
            break

    return G


# ---------------------------------------------------
# Print Mastermind Metrics
# ---------------------------------------------------
from networkx.algorithms.community.quality import modularity

def print_metrics(G, mastermind, stage):

    degree, closeness, betweenness, d_rank, c_rank, b_rank = compute_centrality_and_ranks(G)

    ic_val = estimate_influence(G, mastermind, model="IC")
    lt_val = estimate_influence(G, mastermind, model="LT")

    communities = list(greedy_modularity_communities(G))
    mod_val = modularity(G, communities)

    print("\n====================================")
    print(stage)
    print("====================================")
    print("Mastermind:", mastermind)

    print("\nCentrality Values:")
    print("Degree:", degree[mastermind])
    print("Closeness:", closeness[mastermind])
    print("Betweenness:", betweenness[mastermind])

    print("\nCentrality Ranks:")
    print("Degree Rank:", d_rank[mastermind])
    print("Closeness Rank:", c_rank[mastermind])
    print("Betweenness Rank:", b_rank[mastermind])

    print("\nInfluence Spread:")
    print("IC Influence:", ic_val)
    print("LT Influence:", lt_val)

    print("\nStructural Property:")
    print("Modularity:", mod_val)


def run_experiment(filepath, budget=3, executions=5, simulations=500):

    G = nx.read_edgelist(filepath, nodetype=int)

    # Identify mastermind
    degree, closeness, betweenness, d_rank, c_rank, b_rank = compute_centrality_and_ranks(G)

    sum_ranks = {
        node: d_rank[node] + c_rank[node] + b_rank[node]
        for node in G.nodes()
    }

    mastermind = min(sum_ranks, key=sum_ranks.get)

    print("Mastermind:", mastermind)

    # ---- INITIAL INFLUENCE ----
    ic_initial = estimate_influence(G, mastermind, model="IC", simulations=simulations)
    lt_initial = estimate_influence(G, mastermind, model="LT", simulations=simulations)

    print("\nInitial IC:", ic_initial)
    print("Initial LT:", lt_initial)

    ic_values = [ic_initial]
    lt_values = [lt_initial]

    # ---- Apply ROAM repeatedly ----
    for i in range(executions):
        G = cccr(G, mastermind, budget)

        ic_val = estimate_influence(G, mastermind, model="IC", simulations=simulations)
        lt_val = estimate_influence(G, mastermind, model="LT", simulations=simulations)

        ic_values.append(ic_val)
        lt_values.append(lt_val)

        print(f"\nAfter execution {i+1}:")
        print("IC:", ic_val)
        print("LT:", lt_val)

    # ---- Compute changes ----
    ic_total_change = ic_values[-1] - ic_initial
    lt_total_change = lt_values[-1] - lt_initial

    ic_avg_change = ic_total_change / executions
    lt_avg_change = lt_total_change / executions

    print("\n====================================")
    print("SUMMARY AFTER", executions, "EXECUTIONS")
    print("====================================")

    print("Total IC Change:", ic_total_change)
    print("Average IC Change per execution:", ic_avg_change)

    print("Total LT Change:", lt_total_change)
    print("Average LT Change per execution:", lt_avg_change)
    
run_experiment("facebook_fragment_61.txt", budget=3, executions=3)

# executions here means - “If I repeatedly apply ROAM 20 times to structurally dilute the mastermind's neighbourhood,
# how does their influence change over time?”
```

```python
import networkx as nx
import random
import numpy as np
from networkx.algorithms.community import louvain_communities
from networkx import degree_centrality, closeness_centrality, betweenness_centrality

random.seed(42)
np.random.seed(42)


# -----------------------------------------------------------------------
# BUG FIX 1 — Independent Cascade
# ORIGINAL: used 1/G.degree(node) as activation probability.
#   Problem: degree changes each time ROAM adds/removes edges, so the
#   probability shifts after every iteration, making IC artificially
#   volatile and unrelated to the paper's model.
# FIX: use the paper-specified FIXED probability p = 0.15 on every edge.
# -----------------------------------------------------------------------
def independent_cascade(G, source, p=0.15, simulations=2000):
    spreads = []
    for _ in range(simulations):
        active = {source}
        newly_active = {source}
        while newly_active:
            next_active = set()
            for node in newly_active:
                for neighbor in G.neighbors(node):
                    if neighbor not in active and random.random() <= p:
                        next_active.add(neighbor)
            newly_active = next_active
            active |= newly_active
        spreads.append(len(active))
    return np.mean(spreads) / G.number_of_nodes()


# -----------------------------------------------------------------------
# BUG FIX 2 — Linear Threshold
# ORIGINAL: used influence weight 1/G.degree(neigh) with thresholds 0.1–0.5.
#   Problem: those thresholds are far too high relative to the cumulative
#   influence weights (which sum to 1.0 over all neighbors), so almost no
#   node ever activates after the direct seed step. This produces a flat
#   near-zero LT trace that cannot show the paper's increasing trend.
# FIX: use the paper's integer count model — threshold tv sampled uniformly
#   from {0, …, degree(v)} and activation fires when COUNT of active
#   neighbors ≥ tv.  This matches the Supporting Information of Waniek
#   et al. (2018) exactly.
# -----------------------------------------------------------------------
def linear_threshold(G, source, simulations=2000):
    spreads = []
    for _ in range(simulations):
        # Re-sample thresholds each simulation (paper: "uniform distribution")
        thresholds = {
            node: random.randint(0, max(1, G.degree(node)))
            for node in G.nodes()
        }
        active = {source}
        while True:
            new_active = set()
            for node in G.nodes():
                if node not in active:
                    active_neighbor_count = sum(
                        1 for nb in G.neighbors(node) if nb in active
                    )
                    if active_neighbor_count >= thresholds[node]:
                        new_active.add(node)
            if not new_active:
                break
            active |= new_active
        spreads.append(len(active))
    return np.mean(spreads) / G.number_of_nodes()


# -----------------------------------------------------------------------
# Centrality helpers
# -----------------------------------------------------------------------
def compute_centrality_and_ranks(G):
    deg  = degree_centrality(G)
    clos = closeness_centrality(G)
    betw = betweenness_centrality(G)

    def rank(d):
        return {n: r + 1 for r, n in enumerate(
            sorted(d, key=d.get, reverse=True))}

    return deg, clos, betw, rank(deg), rank(clos), rank(betw)


def identify_mastermind(G):
    _, _, _, dr, cr, br = compute_centrality_and_ranks(G)
    return min(G.nodes(), key=lambda n: dr[n] + cr[n] + br[n])


# -----------------------------------------------------------------------
# BUG FIX 3 — ROAM (paper-faithful implementation)
# ORIGINAL: Step 2 connected v0 to ANY low-degree nodes across the whole
#   graph.  This is wrong.  Per Waniek et al., Step 2 must connect v0
#   only to nodes that are NEIGHBOURS OF THE MASTERMIND (and not yet
#   connected to v0).  This is the structural move that:
#     (a) creates indirect mastermind→v0 paths (preserving LT influence), and
#     (b) dilutes mastermind's relative centrality (v0's new neighbours are
#         already close to the mastermind, so they "share" its position).
#   Connecting v0 to random distant nodes does neither.
# -----------------------------------------------------------------------
def roam(G, mastermind, budget):
    """
    ROAM heuristic (Waniek et al. 2018):
      Step 1 — Remove edge (mastermind, v0) where v0 is mastermind's
               highest-degree neighbour.
      Step 2 — Connect v0 to (budget-1) of mastermind's *remaining*
               neighbours with the lowest degree (not already connected to v0).
    """
    G = G.copy()
    neighbors = list(G.neighbors(mastermind))
    if len(neighbors) < 2:
        return G

    # Step 1: pick and disconnect the highest-degree neighbour
    v0 = max(neighbors, key=lambda x: G.degree(x))
    G.remove_edge(mastermind, v0)

    # Step 2: reconnect v0 to budget-1 of mastermind's remaining neighbours
    #   — MUST be from mastermind's neighbourhood, not arbitrary nodes.
    candidates = sorted(
        [n for n in G.neighbors(mastermind) if not G.has_edge(v0, n)],
        key=lambda x: G.degree(x)   # lowest-degree first (paper's choice)
    )
    for node in candidates[:budget - 1]:
        G.add_edge(v0, node)

    return G


# -----------------------------------------------------------------------
# HERMES heuristic (Saha et al. 2025)
# -----------------------------------------------------------------------
def identify_core_community(G):
    """Return the community with the lowest average centrality rank."""
    comms = louvain_communities(G, seed=42)
    _, _, _, dr, cr, br = compute_centrality_and_ranks(G)
    best_comm, best_avg = None, float('inf')
    for comm in comms:
        avg = np.mean([(dr[n] + cr[n] + br[n]) / 3.0 for n in comm])
        if avg < best_avg:
            best_avg = avg
            best_comm = set(comm)
    return best_comm


def hermes(G, mastermind, budget, d_param=2):
    """
    HERMES heuristic:
      1. Identify the core community containing the mastermind.
      2. Find the top-d most influential EXTERNAL neighbours of the mastermind
         (neighbours outside the core community).
      3. Remove the mastermind's direct links to those external nodes.
      4. Reconnect each external node to a less-influential core member.

    Key insight: mastermind's CORE connections are preserved; only external
    ties are severed.  The newly added core→external edges create new
    indirect influence paths that tend to increase LT spread while
    reducing the mastermind's visible centrality.
    """
    G = G.copy()
    core = identify_core_community(G)

    # External neighbours of mastermind (outside core)
    external = [n for n in G.neighbors(mastermind) if n not in core]
    if not external:
        return G

    _, _, _, dr, cr, br = compute_centrality_and_ranks(G)

    # Top-d most influential external nodes
    top_d_external = sorted(
        external, key=lambda n: dr[n] + cr[n] + br[n]
    )[:d_param]

    # Least influential core members (excluding mastermind itself)
    core_others = sorted(
        [n for n in core if n != mastermind],
        key=lambda n: dr[n] + cr[n] + br[n],
        reverse=True          # highest rank number = least influential
    )

    reconnect_idx = 0
    for ext_node in top_d_external:
        # Sever mastermind → external link
        if G.has_edge(mastermind, ext_node):
            G.remove_edge(mastermind, ext_node)

        # Connect external node to a less-influential core member
        while reconnect_idx < len(core_others):
            target = core_others[reconnect_idx]
            reconnect_idx += 1
            if not G.has_edge(ext_node, target):
                G.add_edge(ext_node, target)
                break

    return G


# -----------------------------------------------------------------------
# Experiment runner
# -----------------------------------------------------------------------
def run_experiment(filepath, budget=3, d_param=2, executions=3, simulations=500):
    G_orig = nx.read_edgelist(filepath, nodetype=int)
    mastermind = identify_mastermind(G_orig)

    print(f"Network : {filepath}")
    print(f"Nodes   : {G_orig.number_of_nodes()}  |  Edges: {G_orig.number_of_edges()}")
    print(f"Mastermind node : {mastermind}  (degree {G_orig.degree(mastermind)})")
    print(f"Budget b={budget}, d={d_param}, simulations={simulations}\n")

    header = f"{'Iter':<5} {'ROAM-IC':>10} {'ROAM-LT':>10} {'HERMES-IC':>12} {'HERMES-LT':>12}"
    print(header)
    print("-" * len(header))

    G_roam   = G_orig.copy()
    G_hermes = G_orig.copy()

    ic0 = independent_cascade(G_orig, mastermind, simulations=simulations)
    lt0 = linear_threshold(G_orig, mastermind, simulations=simulations)
    print(f"{'Init':<5} {ic0:>10.4f} {lt0:>10.4f} {ic0:>12.4f} {lt0:>12.4f}")

    for i in range(1, executions + 1):
        G_roam   = roam(G_roam, mastermind, budget)
        G_hermes = hermes(G_hermes, mastermind, budget, d_param)

        r_ic = independent_cascade(G_roam,   mastermind, simulations=simulations)
        r_lt = linear_threshold(G_roam,      mastermind, simulations=simulations)
        h_ic = independent_cascade(G_hermes, mastermind, simulations=simulations)
        h_lt = linear_threshold(G_hermes,    mastermind, simulations=simulations)

        print(f"{i:<5} {r_ic:>10.4f} {r_lt:>10.4f} {h_ic:>12.4f} {h_lt:>12.4f}")

    print()
    print("Expected trend  (Table 2, HERMES paper):")
    print("  IC  →  decreasing after each iteration")
    print("  LT  →  increasing after each iteration")
    print()
    print("Note: The visibility of the trend depends on the network size and")
    print("density. On larger / sparser networks (e.g., full 4039-node Facebook)")
    print("the effect is more pronounced. On small dense graphs the change per")
    print("iteration is small; use more executions to see the trend clearly.")


# -----------------------------------------------------------------------
# Run
# -----------------------------------------------------------------------
run_experiment(
    "facebook_fragment_333.txt",
    budget=3,
    d_param=2,
    executions=3,
    simulations=2000   # increase to 2000+ for less noisy estimates
)

# NOTE on executions:
# The paper's Figure 3 plots 10-20 consecutive ROAM executions.
# With only 3 executions the per-step change (~0.01) can be within
# simulation noise. Increase executions=10 and simulations=2000 for
# cleaner trends.
```

```python
import networkx as nx
import random
import numpy as np
from networkx.algorithms.community import louvain_communities

random.seed(42)
np.random.seed(42)


# ---------------------------------------------------
# Rank nodes inside a community (random tie breaking)
# ---------------------------------------------------
def rank_nodes(G, nodes):

    sub = G.subgraph(nodes)

    dc = nx.degree_centrality(sub)
    cc = nx.closeness_centrality(sub)
    bc = nx.betweenness_centrality(sub)

    def rank_metric(metric):

        items = list(metric.items())

        random.shuffle(items)  # break ties randomly

        sorted_nodes = sorted(items, key=lambda x: x[1], reverse=True)

        return {node: i+1 for i, (node, _) in enumerate(sorted_nodes)}

    dc_rank = rank_metric(dc)
    cc_rank = rank_metric(cc)
    bc_rank = rank_metric(bc)

    return dc_rank, cc_rank, bc_rank


# ---------------------------------------------------
# Find core community
# ---------------------------------------------------
def find_core_community(G, communities):

    best_avg = float("inf")
    core = None

    for community in communities:

        dc_rank, cc_rank, bc_rank = rank_nodes(G, community)

        total = 0

        for node in community:
            total += dc_rank[node] + cc_rank[node] + bc_rank[node]

        avg_rank = total / len(community)

        if avg_rank < best_avg:
            best_avg = avg_rank
            core = community

    return core


# ---------------------------------------------------
# Find mastermind
# ---------------------------------------------------
def find_mastermind(G, core):

    dc_rank, cc_rank, bc_rank = rank_nodes(G, core)

    scores = {}

    for node in core:

        scores[node] = dc_rank[node] + cc_rank[node] + bc_rank[node]

    mastermind = min(scores, key=scores.get)

    return mastermind


# ---------------------------------------------------
# Find weakest node in a community
# ---------------------------------------------------
def weakest_node(G, community):

    sub = G.subgraph(community)

    dc = nx.degree_centrality(sub)

    weakest = min(dc, key=dc.get)

    return weakest


# ---------------------------------------------------
# Compute global centrality ranks
# ---------------------------------------------------
def global_ranks(G):

    dc = nx.degree_centrality(G)
    cc = nx.closeness_centrality(G)
    bc = nx.betweenness_centrality(G)

    def rank(metric):

        items = list(metric.items())

        random.shuffle(items)

        sorted_nodes = sorted(items, key=lambda x: x[1], reverse=True)

        return {node: i+1 for i,(node,_) in enumerate(sorted_nodes)}

    return rank(dc), rank(cc), rank(bc)


# ---------------------------------------------------
# C3R rewiring
# ---------------------------------------------------
def c3r_rewire(G, mastermind, core, communities, d):

    G = G.copy()

    external_nodes = [n for n in G.neighbors(mastermind) if n not in core]

    if len(external_nodes) == 0:
        return G

    # global centrality ranking
    dc_rank, cc_rank, bc_rank = global_ranks(G)

    scores = {}

    for n in external_nodes:
        scores[n] = dc_rank[n] + cc_rank[n] + bc_rank[n]

    # top d influential external nodes
    top_external = sorted(scores, key=scores.get)[:d]

    for Oi in top_external:

        # find Oi community
        Oi_comm = None

        for comm in communities:
            if Oi in comm:
                Oi_comm = comm
                break

        if Oi_comm is None:
            continue

        Wi = weakest_node(G, Oi_comm)
        MWi = weakest_node(G, core)

        # remove mastermind → Oi
        if G.has_edge(mastermind, Oi):
            G.remove_edge(mastermind, Oi)

        # add chain edges
        if not G.has_edge(Oi, Wi):
            G.add_edge(Oi, Wi)

        if not G.has_edge(Wi, MWi):
            G.add_edge(Wi, MWi)

        if not G.has_edge(MWi, mastermind):
            G.add_edge(MWi, mastermind)

    return G


# ---------------------------------------------------
# Main C3R Experiment
# ---------------------------------------------------
def run_c3r(filepath, iterations=3, d=2):

    G = nx.read_edgelist(filepath, nodetype=int)

    print("Graph loaded.")
    print("Nodes:", G.number_of_nodes())
    print("Edges:", G.number_of_edges())

    # -------- INITIAL COMMUNITY --------

    communities = louvain_communities(G, seed=42)

    core = find_core_community(G, communities)

    mastermind = find_mastermind(G, core)

    print("\nCore community size:", len(core))
    print("Mastermind:", mastermind)

    # initial centrality ranks
    dc_rank, cc_rank, bc_rank = global_ranks(G)

    print("\nInitial Mastermind Ranks")
    print("Degree Rank:", dc_rank[mastermind])
    print("Closeness Rank:", cc_rank[mastermind])
    print("Betweenness Rank:", bc_rank[mastermind])

    # -------- ITERATIONS --------

    for i in range(iterations):

        print("\nIteration", i+1)

        communities = louvain_communities(G, seed=42)

        core = find_core_community(G, communities)

        G = c3r_rewire(G, mastermind, core, communities, d)

    # -------- FINAL CENTRALITY --------

    dc_rank_f, cc_rank_f, bc_rank_f = global_ranks(G)

    print("\nFinal Mastermind Ranks")
    print("Degree Rank:", dc_rank_f[mastermind])
    print("Closeness Rank:", cc_rank_f[mastermind])
    print("Betweenness Rank:", bc_rank_f[mastermind])

    return G


# ---------------------------------------------------
# Run
# ---------------------------------------------------
run_c3r("facebook_fragment_333.txt", iterations=3, d=2)
```
